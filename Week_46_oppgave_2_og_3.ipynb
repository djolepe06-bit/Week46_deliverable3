## Task 2: Dice class

```python
import random


class Dice:
    """A class representing an n-sided dice."""

    def __init__(self, sides: int = 6) -> None:
        if sides < 2:
            raise ValueError("A dice must have at least 2 sides.")
        self.sides = sides

    def roll(self) -> int:
        """Rolls the dice and returns a random integer between 1 and sides."""
        return random.randint(1, self.sides)

    def __repr__(self) -> str:
        """Returns string representation of the Dice object."""
        return f"Dice(sides={self.sides})"

    def __call__(self) -> int:
        """Allows rolling the dice by calling the instance directly, e.g. d20()."""
        return self.roll()
```

```python
d20 = Dice(sides=20)
print(d20)
print(d20())
```

## Task 3: Create items
In the assignment the `Weapon` class lives in `items.py`. Here it is in the notebook so everything runs in one place.

```python
class Weapon:
    """A weapon that uses a Dice for damage calculations."""

    def __init__(self, name: str, damage_dice: Dice) -> None:
        self.name = name
        self.damage_dice = damage_dice

    def roll_damage(self) -> int:
        """Rolls the weapon's dice and returns the damage."""
        return self.damage_dice.roll()

    def __repr__(self) -> str:
        return f"Weapon(name={self.name!r}, damage_dice={self.damage_dice!r})"
```

```python
class Character:
    def __init__(self, name: str, hp: int = 100, base_damage: int = 10,
                 weapon: Weapon | None = None) -> None:
        self.name = name
        self._hp = hp
        self.base_damage = base_damage
        self.weapon = weapon  # optional weapon

    @property
    def is_alive(self) -> bool:
        return self._hp > 0

    def take_damage(self, damage: int) -> None:
        self._hp = max(0, self._hp - damage)

    def heal(self, amount: int) -> None:
        self._hp += amount

    def attack(self, target: "Character") -> None:
        # Use the weapon's damage if equipped, otherwise default damage
        if self.weapon is not None:
            damage = self.weapon.roll_damage()
        else:
            damage = self.base_damage
        target.take_damage(damage)
```

```python
sword = Weapon("Sword", Dice(8))
hero = Character("Hero", weapon=sword)
unarmed = Character("Peasant")
goblin = Character("Goblin")

hero.attack(goblin)
print(goblin._hp)  # 92 to 99 (d8 damage)

unarmed.attack(goblin)
print(goblin._hp)  # 10 less than before (default damage)
```
